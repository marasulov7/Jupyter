# Депозитный калькулятор

```python
def calculate_deposit(initial_amount, term_years, interest_rate, capitalization):
    periods = {"monthly": 12, "quarterly": 4, "yearly": 1}

    if capitalization not in periods:
        raise ValueError("Неверный срок капитализации. Доступные варианты: monthly, quarterly, yearly.")

    n = periods[capitalization]  # Количество периодов в году
    r = interest_rate / 100  # Процент в десятичной форме

    final_amount = initial_amount * (1 + r / n) ** (n * term_years)
    return round(final_amount, 2)

# Ввод данных от пользователя
initial_amount = float(input("Введите начальную сумму вклада: "))
term_years = int(input("Введите срок вклада (в годах): "))
interest_rate = float(input("Введите процентную ставку: "))
capitalization = input("Выберите капитализацию (monthly, quarterly, yearly): ")

# Расчет и вывод результата
try:
    result = calculate_deposit(initial_amount, term_years, interest_rate, capitalization)
    print(f"Сумма на депозите после {term_years} лет: {result} тенге")
except ValueError as e:
    print(e)
```
