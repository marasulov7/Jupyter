```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

# Функция для отображения текста точками
def text_to_points(text, font_size=100):
    fig, ax = plt.subplots()
    ax.text(0.5, 0.5, text, fontsize=font_size, ha='center', va='center', alpha=0.5)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis('off')
    fig.canvas.draw()
    
    # Получение пиксельных данных
    data = np.array(fig.canvas.renderer.buffer_rgba())[:, :, 0]
    plt.close(fig)
    
    # Преобразование изображения в координаты точек
    points = np.column_stack(np.where(data < 128))
    points = points / np.max(points, axis=0)  # Нормализация
    return points

# Получение координат точек для текста "Алишер"
points = text_to_points("Алишер")

# Визуализация точек
fig, ax = plt.subplots()
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')
ax.scatter(points[:, 1], 1 - points[:, 0], color='black', s=5)
plt.show()
```
