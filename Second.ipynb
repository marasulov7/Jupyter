# Курс доллара к тенге за 2024 год

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Загрузка данных
df = pd.read_csv("usd_kzt_2024.csv")

# Преобразование колонки Date в формат datetime
df["Date"] = pd.to_datetime(df["Date"])

# Создание графика
plt.figure(figsize=(12, 6))
sns.lineplot(x="Date", y="USD_KZT", data=df, color="blue")

# Настройка заголовков
plt.title("Курс доллара к тенге в 2024 году", fontsize=14)
plt.xlabel("Дата", fontsize=12)
plt.ylabel("Курс USD/KZT", fontsize=12)
plt.xticks(rotation=45)
plt.grid(True)

# Показать график
plt.show()
```
